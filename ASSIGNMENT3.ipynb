```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("WhatsApp Image 2026-04-04 at 17.45.39.jpeg")
plt.imshow(image)
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)
```

```python
border_width = int(input("Enter border width in pixels: "))
border_color = int(input("Enter color: "))

gray_border = cv2.copyMakeBorder(
    gray,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

binary_border = cv2.copyMakeBorder(
    binary,
    border_width,
    border_width,
    border_width,
    border_width,
    cv2.BORDER_CONSTANT,
    value=border_color
)

#plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray_border, cmap="gray")
plt.title("Grayscale with Border")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary_border, cmap="gray")
plt.title("Binary with Border")
plt.axis("off")

plt.show()
```

```python
gray_complement = cv2.bitwise_not(gray)
complement = cv2.bitwise_not(binary)
```
